```python
import tkinter as tk
from tkinter import messagebox

# Create main window
root = tk.Tk()
root.title("To-Do List Application")
root.geometry("500x550")
root.configure(bg="#EAF2F8")
root.resizable(False, False)

# Title
title = tk.Label(
    root,
    text="MY TO-DO LIST",
    font=("Arial", 22, "bold"),
    bg="#1F4E79",
    fg="white",
    pady=15
)
title.pack(fill="x")

# Main frame
frame = tk.Frame(root, bg="white", padx=25, pady=20)
frame.pack(padx=25, pady=25, fill="both", expand=True)

# Task label
tk.Label(
    frame,
    text="Enter Your Task",
    font=("Arial", 12, "bold"),
    bg="white"
).pack(anchor="w")

# Task input
task_entry = tk.Entry(
    frame,
    font=("Arial", 12),
    width=35
)
task_entry.pack(pady=10)

# Task list
task_list = tk.Listbox(
    frame,
    font=("Arial", 12),
    width=40,
    height=12,
    selectbackground="#3498DB"
)
task_list.pack(pady=10)

# Add task function
def add_task():
    task = task_entry.get().strip()

    if task:
        task_list.insert(tk.END, task)
        task_entry.delete(0, tk.END)
    else:
        messagebox.showwarning(
            "Warning",
            "Please enter a task!"
        )

# Delete task function
def delete_task():
    selected = task_list.curselection()

    if selected:
        task_list.delete(selected[0])
    else:
        messagebox.showwarning(
            "Warning",
            "Please select a task!"
        )

# Mark completed function
def complete_task():
    selected = task_list.curselection()

    if selected:
        index = selected[0]
        task = task_list.get(index)

        if not task.startswith("✓ "):
            task_list.delete(index)
            task_list.insert(index, "✓ " + task)

    else:
        messagebox.showwarning(
            "Warning",
            "Please select a task!"
        )

# Clear all tasks
def clear_tasks():
    if task_list.size() > 0:
        task_list.delete(0, tk.END)

# Buttons
tk.Button(
    frame,
    text="ADD TASK",
    command=add_task,
    bg="#2980B9",
    fg="white",
    font=("Arial", 11, "bold"),
    width=20
).pack(pady=5)

tk.Button(
    frame,
    text="MARK COMPLETED",
    command=complete_task,
    bg="#27AE60",
    fg="white",
    font=("Arial", 11, "bold"),
    width=20
).pack(pady=5)

tk.Button(
    frame,
    text="DELETE TASK",
    command=delete_task,
    bg="#E67E22",
    fg="white",
    font=("Arial", 11, "bold"),
    width=20
).pack(pady=5)

tk.Button(
    frame,
    text="CLEAR ALL",
    command=clear_tasks,
    bg="#C0392B",
    fg="white",
    font=("Arial", 11, "bold"),
    width=20
).pack(pady=5)

# Footer
tk.Label(
    root,
    text="Developed using Python Tkinter",
    font=("Arial", 9),
    bg="#EAF2F8",
    fg="#555555"
).pack(pady=5)

# Run application
root.mainloop()
```
